```python
antigravity_cli: AntigravityCLIProvider = AntigravityCLIProvider()
```

```python
await antigravity_cli.is_authenticated()
```

```python
models = await antigravity_cli.get_models()
models
```

```python
model_id: str = "gemini-3.8-flash-low"
model = await antigravity_cli.get_model(model_id)
```

```python
thinking_levels = await model.get_thinking_levels()
print(thinking_levels)
```

```python
session = await antigravity_cli.create_session(
    model=model,
    workspace_id="/workspace",
)
```

```python
from core.models.run import RunRequest

run = await antigravity_cli.run(
    session,
    RunRequest(prompt="What is my name?"),
)
```

```python
print(run.result)
print(run.status)
```
