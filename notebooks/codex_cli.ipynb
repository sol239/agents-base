```python
codex_cli: CodexCLIProvider = CodexCLIProvider()
```

```python
await codex_cli.is_authenticated()
```

```python
models = await codex_cli.get_models()
models
```

```python
model_id: str = "gpt-6-luna"
luna = await codex_cli.get_model(model_id)
if luna is None:
    raise ValueError(f"Model not found: {model_id}")
```

```python
luna.thinking_level = "low"
```

```python
session = await codex_cli.create_session(
    model=luna,
    workspace_id="/workspace",
)
```

```python
from core.models.run import RunRequest

run = await codex_cli.run(
    session,
    RunRequest(prompt="Hello!"),
)
```

```python
print(run.result)
print(run.status)
```
